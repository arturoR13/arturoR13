# exercise 1: router status monitoring for loop.

```python
routers=["192.168.1.1","192.186.1.2","192.186.1.3","192.186.1.4"]
for router in routers:
      if router.endswith(".3"):
          status = "offline"
      else:
            status = "online"
      print(router,status)
```

# exercise 2: dynamic router list while loop

```python
routers = []
while True:
      ip = input("enter router ip(or 'done' to finish):")
      if ip.lower()=="done":
            break
      routers.append(ip)
print("\nrouters registered:")

for router in routers:
      print(router)
```

# exercise 3 : router status monitoring_using dicctionaries


```python
router_status = {}
while True:
      ip = input ("enter router ip or 'done' to finish:")
      if ip.lower() == "done":
              break
      router_status[ip]="online"
for ip, status in router_status.items():
  print(f"ip: {ip} - status: {status}")
```
